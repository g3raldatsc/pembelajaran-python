      <h2 style="margin-top: 0;">Pengantar Dasar Sintaks Python</h3>
      <p style="text-align: justify; margin-bottom: 10px;">
        Sebelum kamu mulai menulis baris kode pertamamu, kamu perlu memahami apa itu sintaks. Secara sederhana, sintaks adalah tata bahasa atau aturan penulisan yang harus dipatuhi agar instruksi yang kamu berikan dapat dimengerti oleh komputer. Jika tata bahasanya salah, komputer tidak akan bisa menerjemahkan keinginanmu, dan program akan berhenti bekerja.
      </p>
      <p style="text-align: justify; margin-bottom: 10px;">
        Python dirancang dengan filosofi bahwa kode program harus mudah dibaca, bahkan oleh mereka yang baru belajar. Berbeda dengan bahasa pemrograman lain yang sering mewajibkan penggunaan simbol rumit (seperti titik koma di akhir baris atau kurung kurawal untuk membungkus blok kode), Python menggunakan pendekatan yang lebih bersih. Di sini, pergantian baris (Enter) menandakan selesainya sebuah perintah. Ini membuatmu bisa lebih fokus pada logika penyelesaian masalah daripada aturan penulisan simbol.
      </p>
      <p style="text-align: justify; margin-bottom: 10px;">
        Hal yang harus kamu ingat adalah sifat <strong>Case Sensitive</strong>. Python membedakan secara tegas antara huruf besar dan huruf kecil. Sebagai contoh, variabel dengan nama <code style="color: #000000; background-color: #f0f0f0; padding: 2px 4px; border-radius: 3px;">Data</code>, <code style="color: #000000; background-color: #f0f0f0; padding: 2px 4px; border-radius: 3px;">DATA</code>, dan <code style="color: #000000; background-color: #f0f0f0; padding: 2px 4px; border-radius: 3px;">data</code> dianggap sebagai tiga hal yang sepenuhnya berbeda. Ketidaktelitian dalam penulisan huruf ini adalah penyebab paling umum terjadinya kesalahan (error) saat kamu baru mulai belajar.
      </p>
      <p style="text-align: justify; margin-bottom: 10px;">
        Mekanisme eksekusi kode di Python berjalan secara berurutan dari baris paling atas ke bawah (top-down). Interpreter Python akan membaca dan menjalankan kodemu baris demi baris. Jika terdapat satu kesalahan sintaks di tengah jalan, proses akan langsung terhenti dan memunculkan pesan <code style="color: #000000; background-color: #f0f0f0; padding: 2px 4px; border-radius: 3px;">SyntaxError</code>. Oleh karena itu, ketelitian dalam mengikuti aturan dasar ini adalah kunci utama keberhasilan programmu.
      </p>

        <h2 style="margin: 0; color: #000000; font-family: sans-serif;">Komentar: Seni Mendokumentasikan Kode</h3>
        <p style="margin-top: 5px; color: #444;">Kode ditulis untuk komputer, tetapi komentar ditulis untuk manusia.</p>
      </div>
      <p style="text-align: justify; margin-bottom: 15px; line-height: 1.6;">
        Setelah memahami sintaks dan indentasi, elemen vital berikutnya adalah <strong>Komentar</strong>. Komentar adalah baris teks dalam kodemu yang sepenuhnya diabaikan oleh interpreter Python. Saat program dijalankan, komputer akan melewati baris ini seolah-olah mereka tidak ada.
      </p>
      <p style="text-align: justify; margin-bottom: 20px; line-height: 1.6;">
        Fungsi utama komentar bukan untuk mempengaruhi jalannya program, melainkan sebagai jejak dokumentasi. Ini berguna untuk menjelaskan logika yang rumit, memberikan catatan kepada programmer lain (atau dirimu sendiri di masa depan), serta mematikan sementara baris kode tertentu saat proses perbaikan bug (debugging).
      </p>
      <div style="background-color: #f9f9f9; border-left: 6px solid #ffd343; padding: 20px; margin-bottom: 25px;">
        <h4 style="margin-top: 0; color: #000000; margin-bottom: 15px;">Dua Jenis Komentar di Python</h4>        
        <div style="margin-bottom: 15px;">
          <strong style="color: #000000;">1. Komentar Satu Baris (Single-line)</strong><br>
          <p style="margin: 5px 0;">Jenis ini diawali dengan tanda pagar (<code style="color: #000000; background-color: #e0e0e0; padding: 2px 5px; border-radius: 3px;">#</code>). Semua teks setelah tanda ini hingga akhir baris akan dianggap sebagai komentar.</p>
        </div>
        <div>
          <strong style="color: #000000;">2. Komentar Banyak Baris (Multi-line)</strong><br>
          <p style="margin: 5px 0;">Python menggunakan tanda kutip tiga (tiga kutip satu <code style="color: #000000; background-color: #e0e0e0; padding: 2px 5px; border-radius: 3px;">'''</code> atau tiga kutip ganda <code style="color: #000000; background-color: #e0e0e0; padding: 2px 5px; border-radius: 3px;">"""</code>) untuk membungkus paragraf panjang. Teknik ini sering disebut sebagai <em>Docstring</em>.</p>
        </div>
      </div>
      <h3 style="color: #000000; margin-top: 0;">Praktik Terbaik</h3>
      <p style="text-align: justify; margin-bottom: 10px; line-height: 1.6;">
        Ada perbedaan besar antara komentar yang baik dan komentar yang buruk. Jangan menulis komentar untuk menjelaskan <em>apa</em> yang dilakukan kode jika itu sudah jelas terlihat, tetapi jelaskanlah <em>mengapa</em> kamu menulis kode tersebut.
      </p>
      <div style="display: flex; gap: 20px; margin-top: 10px;">
        <div style="flex: 1; background-color: #ffe6e6; padding: 15px; border-radius: 5px; border: 1px solid #ffcccc;">
          <strong style="color: #cc0000; display: block; margin-bottom: 10px;">Contoh Kurang Tepat</strong>
          <code style="display: block; font-family: monospace; color: #000000;">
            x = 10  # Mengisi variabel x dengan 10
          </code>
          <p style="font-size: 13px; color: #444; margin-top: 5px;">(Ini berlebihan karena kodenya sudah jelas)</p>
        </div>       
        <div style="flex: 1; background-color: #e6fffa; padding: 15px; border-radius: 5px; border: 1px solid #b3f0e0;">
          <strong style="color: #006644; display: block; margin-bottom: 10px;">Contoh Tepat</strong>
          <code style="display: block; font-family: monospace; color: #000000;">
            x = 10  # Batas maksimal percobaan login
          </code>
          <p style="font-size: 13px; color: #444; margin-top: 5px;">(Menjelaskan konteks atau alasan)</p>
        </div>
      </div>

      <div style="text-align: justify; color: #333;">
        <p style="margin-bottom: 10px;">
          Secara umum, Variabel merupakan tempat atau wadah untuk menyimpan nilai atau data tertentu yang disimpan di dalam memori komputer yang memungkinkan kita untuk mengaksesnya terus-menerus, mengubahnya, dan dimanipulasi. Nilai variabel, harus memiliki tipe data tertentu (yang akan kita pelajari di materi selanjutnya).
        </p>
        <p style="margin-bottom: 10px;">
          Proses pembuatan variabel biasanya melalui tahap deklarasi (pengenalan variabel) dan inisialisasi (pemberian nilai awal).
        </p>
        <p style="margin-bottom: 10px;">
          Mudahnya, anggap kamu mempunyai sebuah toples kosong. Toples kosong itu kita bisa sebut sebagai variabel (wadah). Kamu ingin memasukkan snack Doritoz di dalam wadah tersebut. Snack Doritoz itu bisa kita sebut sebagai nilai dari variabel (isi wadah).
        </p>
        <p style="margin-bottom: 10px;">
          Nah, dengan analogi itu, snack Doritoz tersebut sudah tersimpan aman di dalam toples yang awalnya kosong. Jika kita mengambil beberapa keping dari Doritoz tersebut, maka jumlah Doritoz tersebut akan berkurang di dalam toples. Itu adalah analogi yang membuktikan bahwa nilai dari variabel bisa diubah dan dimanipulasi.
        </p>
        <p style="margin-bottom: 20px;">
          Anggap lagi kita mengambil satu keping Doritoz dan menandainya dengan spidol berwarna lalu menyimpannya kembali di dalam toples. Ketika kita mau ambil Doritoz yang sudah ditandai dengan spidol berwarna, maka kita tinggal mengambilnya di toples kapanpun kita mau karena kita memang sudah menyimpannya di dalam sana. Hal itu merupakan analogi bahwa variabel bisa diakses kapanpun.
        </p>
      </div>

</br>toples = "Doritoz (Sisa Setengah)"
</br># 3. Akses (Mengambil Doritoz bertanda spidol)
# Kita bisa memanggil variabel 'toples' kapanpun
</br>print(toples)</pre>

      <h3 style="margin-top: 0; color: #000; margin-bottom: 20px;">Variabel Di Python</h3>     
      <div style="text-align: justify; color: #333; line-height: 1.6;">       
        <p style="margin-bottom: 15px;">
          Jika kamu menganggap bahwa variabel itu seperti wadah, kamu tidaklah salah. Akan tetapi, di Python anggapan tersebut kurang tepat.
        </p>
        <div style="border-left: 4px solid #d1d5da; padding-left: 15px; margin-bottom: 20px; color: #586069;">
          <p style="margin: 0;">
            Dalam pemrograman Python, variabel bukanlah sebuah "wadah" atau "kotak" statis tempat kita bisa menyimpan data (seperti konsep pada bahasa pemrograman low-level).
          </p>
        </div>
        <p style="margin-bottom: 15px;">
          Secara definisi teknis di Python, variabel adalah sebuah nama simbolis (label) yang mereferensikan atau menunjuk ke sebuah objek (nilai) di dalam memori komputer.
        </p>
        <p style="margin-bottom: 10px; font-weight: bold;">
          Ketika kita membuat variabel, Anda sebenarnya sedang melakukan dua hal:
        </p>
        <ol style="margin-bottom: 20px; padding-left: 20px;">
            <li style="margin-bottom: 5px;">Membuat sebuah objek (nilai) di lokasi memori tertentu.</li>
            <li>Memberikan nama (label/variabel) yang menunjuk ke lokasi memori tersebut.</li>
        </ol>
        <p style="margin-bottom: 15px;">
          Intinya, kita itu hanya memberi nama untuk suatu nilai yang telah tersimpan di dalam memori komputer. Simpelnya, Kita beri nama karena nilai yang sudah tersimpan di dalam memori komputer memiliki alamat yang sulit untuk diingat atau diakses. Sehingga, dengan membuat nama dari nilai tersebut, memudahkan kita untuk mengaksesnya dan mengingatnya.
        </p>
        <p style="margin-bottom: 20px;">
          Python merupakan bahasa pemrograman High-Level yang merupakan bahasa yang paling mudah untuk dimengerti oleh manusia atau anggapannya bahasa pemrograman yang mendukung manusia dalam segala aspek.
        </p>
      </div>
      <div style="background-color: #f6f8fa; padding: 15px; border-radius: 5px; border: 1px solid #e1e4e8;">
        <pre style="margin: 0; font-family: Consolas, 'Courier New', monospace; font-size: 12px; color: #24292e; line-height: 1.5;">
# Ilustrasi Variabel sebagai Label (Bukan Wadah)
# 1. Membuat objek (nilai) 100 di memori
# 2. Memberikan label 'nilai_ujian' ke objek tersebut
</br>nilai_ujian = 100
</br># Kita bisa mengecek "alamat rumah" (lokasi memori) dari nilai tersebut
# Ini membuktikan bahwa variabel hanya menunjuk ke alamat ini
</br>print(id(nilai_ujian)) 
</br># Jika kita buat label baru menunjuk ke nilai yang sama
</br>hasil_akhir = nilai_ujian
</br># Maka alamatnya akan sama (karena menunjuk barang yang sama)
</br>print(id(hasil_akhir))</pre>

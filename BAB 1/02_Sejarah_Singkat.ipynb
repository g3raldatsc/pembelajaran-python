      <h3 style="margin-top: 0; color: #000;">Evolusi: Transisi Python 2 ke Python 3</h3>
      <p style="text-align: justify; margin-bottom: 20px; color: #333;">
        Dalam mempelajari sejarah Python, kamu perlu mengetahui satu peristiwa penting yang membagi komunitas pengembang menjadi dua, yaitu transisi dari Python versi 2 ke Python versi 3.
      </p>
      <div style="background-color: #f5f5f5; padding: 15px; border-radius: 10px; border: 1px solid #ddd; margin-bottom: 20px;">
        <h4 style="margin-top: 0; color: #1565c0;">Mengapa Terjadi Perubahan Besar?</h4>
        <p style="text-align: justify; margin-bottom: 10px; font-size: 14px; color: #333;">
          Pada tahun 2008, Guido van Rossum memutuskan untuk merilis Python 3.0 untuk memperbaiki cacat desain fundamental yang ada pada Python 2. Tujuannya adalah membuat bahasa ini lebih logis, konsisten, dan modern.
        </p>
        <p style="text-align: justify; margin-bottom: 0; font-size: 14px; color: #333;">
          <strong>Masalah Kompatibilitas:</strong> Python 3 tidak memiliki fitur <em>backward compatibility</em> dengan Python 2. Artinya, kode program yang ditulis menggunakan sintaks Python 2 tidak dapat dijalankan di Python 3 tanpa modifikasi. Hal ini sempat menyebabkan adopsi Python 3 berjalan lambat di awal peluncurannya.
        </p>
      </div>
      <h4 style="margin-bottom: 10px; color: #000;">Linimasa Penting</h4>
      <div style="margin-left: 10px; border-left: 3px solid #ccc; padding-left: 20px; margin-bottom: 25px;">
        <div style="margin-bottom: 15px;">
          <strong style="color: #333;">1991 - Rilis Pertama</strong><br>
          <span style="color: #555; font-size: 14px;">Python versi 0.9.0 dirilis ke publik oleh Guido van Rossum.</span>
        </div>
        <div style="margin-bottom: 15px;">
          <strong style="color: #333;">2000 - Python 2.0</strong><br>
          <span style="color: #555; font-size: 14px;">Rilis yang memperkenalkan fitur-fitur modern seperti <em>List Comprehensions</em> dan sistem <em>Garbage Collection</em>.</span>
        </div>
        <div style="margin-bottom: 15px;">
          <strong style="color: #1565c0;">2008 - Python 3.0</strong><br>
          <span style="color: #555; font-size: 14px;">Perombakan besar-besaran untuk memperbaiki inkonsistensi bahasa. Mulai saat ini, fokus pengembangan beralih ke versi 3.</span>
        </div>
        <div>
          <strong style="color: #d32f2f;">2020 - Akhir Python 2 (End of Life)</strong><br>
          <span style="color: #555; font-size: 14px;">Dukungan resmi untuk Python 2 dihentikan sepenuhnya. Pengguna sangat disarankan untuk beralih total ke Python 3.</span>
        </div>

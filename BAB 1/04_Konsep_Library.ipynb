      <h3 style="margin-top: 0; color: #000000; margin-bottom: 20px;">Memahami Konsep Library</h3>    
      <div style="display: flex; flex-wrap: wrap; align-items: center; gap: 25px;">      
        <div style="flex: 1; min-width: 300px;">
          <p style="text-align: justify; margin-bottom: 15px; color: #000000;">
            Dalam pengembangan perangkat lunak modern, efisiensi adalah prioritas utama. Hal ini melahirkan prinsip <strong>DRY (Don't Repeat Yourself)</strong>, yang menekankan bahwa penulisan kode yang berulang untuk fungsi yang sama harus dihindari.
          </p>      
          <div style="border-left: 3px solid #000000; padding-left: 15px; margin-bottom: 15px;">
            <p style="text-align: justify; margin-bottom: 5px; color: #000000;">
              <strong>Definisi Umum:</strong>
            </p>
            <p style="text-align: justify; margin-bottom: 0; font-size: 14px; color: #000000;">
              Library adalah koleksi modul kode yang berisi fungsi-fungsi, kelas, dan variabel yang telah ditulis, diuji, dan dioptimalkan oleh pengembang lain. Koleksi ini dikemas sedemikian rupa agar dapat digunakan kembali (<em>reusable</em>) dalam berbagai proyek berbeda tanpa perlu menulis ulang logika dasarnya dari nol.
            </p>
          </div>
          <p style="text-align: justify; margin-bottom: 0; color: #000000;">
            Library diciptakan untuk memisahkan logika kompleks menjadi bagian-bagian kecil yang modular. Dengan demikian, kamu dapat fokus pada logika bisnis spesifik aplikasi yang sedang kamu bangun, sementara tugas-tugas umum (seperti perhitungan matematika kompleks, koneksi database, atau manipulasi file) diserahkan kepada Library yang sudah terbukti stabil.
          </p>
        </div>

      <h3 style="margin-top: 0; color: #000000;">Klasifikasi Library Python</h3>
      <p style="text-align: justify; margin-bottom: 25px; color: #000000;">
        Ekosistem Python terbagi menjadi dua kategori utama berdasarkan sumber kodenya. Pemahaman mengenai klasifikasi ini sangat penting untuk mengetahui apakah kamu perlu melakukan instalasi tambahan atau cukup menggunakan fitur yang sudah tersedia.
      </p>
      <div style="margin-bottom: 30px; border-left: 4px solid #000000; padding-left: 20px;">
        <h4 style="margin-top: 0; margin-bottom: 10px; color: #000000; font-weight: bold; text-transform: uppercase;">1. Standard Library (Pustaka Bawaan)</h4>     
        <p style="text-align: justify; margin-bottom: 10px; color: #000000;">
          Kelompok ini adalah modul yang sudah terintegrasi langsung di dalam instalasi Python. Python menganut filosofi <em>"Batteries Included"</em>, yang berarti bahasa ini </br>dirilis lengkap dengan pustaka standar yang kaya fitur. Kamu dapat langsung menggunakannya (melakukan import) tanpa perlu mengunduh apa pun dari internet.
        </p>        
        <p style="margin-bottom: 5px; color: #000000; font-weight: bold;">Contoh Penggunaan:</p>
        <ul style="margin-top: 0; color: #000000; padding-left: 20px;">
          <li style="margin-bottom: 5px;">
            <code style="background-color: #eee; color: #000; padding: 2px 5px; border-radius: 3px;">os</code> : Untuk interaksi dengan sistem operasi (manajemen file/folder).
          </li>
          <li style="margin-bottom: 5px;">
            <code style="background-color: #eee; color: #000; padding: 2px 5px; border-radius: 3px;">math</code> : Untuk fungsi matematika tingkat lanjut (trigonometri, logaritma).
          </li>
          <li>
            <code style="background-color: #eee; color: #000; padding: 2px 5px; border-radius: 3px;">datetime</code> : Untuk pemrosesan tanggal dan waktu real-time.
          </li>
        </ul>
      </div>
      <div style="margin-bottom: 10px; border-left: 4px solid #000000; padding-left: 20px;">
        <h4 style="margin-top: 0; margin-bottom: 10px; color: #000000; font-weight: bold; text-transform: uppercase;">2. External Library (Pihak Ketiga)</h4>        
        <p style="text-align: justify; margin-bottom: 10px; color: #000000;">
          Kelompok ini adalah pustaka yang dikembangkan oleh komunitas open-source global atau perusahaan teknologi. Kode ini tidak disertakan dalam instalasi dasar </br>Python. Library ini disimpan di repositori publik bernama PyPI (Python Package Index). Untuk menggunakannya, kamu wajib mengunduhnya terlebih dahulu </br>menggunakan <em>package manager</em> bernama <code style="background-color: #eee; color: #000; padding: 2px 5px; border-radius: 3px;">pip</code>.
        </p>       
        <p style="margin-bottom: 5px; color: #000000; font-weight: bold;">Contoh Penggunaan:</p>
        <ul style="margin-top: 0; color: #000000; padding-left: 20px;">
          <li style="margin-bottom: 5px;">
            <code style="background-color: #eee; color: #000; padding: 2px 5px; border-radius: 3px;">pandas</code> : Untuk analisis dan manipulasi data tabel.
          </li>
          <li style="margin-bottom: 5px;">
            <code style="background-color: #eee; color: #000; padding: 2px 5px; border-radius: 3px;">tensorflow</code> : Untuk pengembangan Deep Learning dan AI.
          </li>
          <li>
            <code style="background-color: #eee; color: #000; padding: 2px 5px; border-radius: 3px;">django</code> : Framework lengkap untuk pengembangan server web (Backend).
          </li>
        </ul>
      </div>

      <h3 style="margin-top: 0; color: #000000;">Konsep Import dalam Python</h3>
      <p style="text-align: justify; margin-bottom: 25px; color: #000000;">
        Mungkin timbul pertanyaan, mengapa Python tidak memuat semua library tersebut secara otomatis sejak awal agar pengguna tidak perlu repot mengetik perintah pemanggilan manual. Jawabannya berkaitan erat dengan strategi manajemen sumber daya komputer.
      </p>
      <p style="text-align: justify; margin-bottom: 25px; color: #000000;">
        Python dirancang dengan pendekatan efisiensi tinggi. Berikut adalah dua alasan teknis utama mengapa mekanisme import dibuat secara manual:
      </p>
      <div style="margin-bottom: 25px; border-left: 4px solid #000000; padding-left: 20px;">
        <h4 style="margin-top: 0; margin-bottom: 10px; color: #000000; font-weight: bold; text-transform: uppercase;">1. Menjaga Kecepatan Program</h4>
        <p style="text-align: justify; margin-bottom: 0; color: #000000; line-height: 1.6;">
          Python memiliki ribuan library eksternal untuk berbagai keperluan, mulai dari matematika, antariksa, hingga pembuatan website. Jika Python dipaksa untuk memuat seluruh library tersebut setiap kali kamu menjalankan program sederhana, waktu start-up akan menjadi sangat lama. Dengan sistem import, Python hanya perlu menyiapkan alat yang benar-benar akan digunakan.
        </p>
      </div>
      <div style="margin-bottom: 10px; border-left: 4px solid #000000; padding-left: 20px;">
        <h4 style="margin-top: 0; margin-bottom: 10px; color: #000000; font-weight: bold; text-transform: uppercase;">2. Efisiensi Memori (RAM)</h4>
        <p style="text-align: justify; margin-bottom: 0; color: #000000; line-height: 1.6;">
          Setiap kode yang dimuat akan memakan ruang di Random Access Memory (RAM). Jika semua library dimuat sekaligus, memori komputer akan penuh dengan fitur yang tidak terpakai. Mekanisme import memberikan kendali penuh kepada kamu untuk menentukan beban kerja memori sesuai kebutuhan spesifik proyek yang sedang dikerjakan.
        </p>
      </div>
    </td>

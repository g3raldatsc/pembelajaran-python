      <h3 style="margin-top: 0;">Apa Itu Python?</h2>
      <p style="text-align: justify; margin-bottom: 10px;">
        Python adalah bahasa pemrograman tingkat tinggi yang dirancang khusus agar mudah dibaca dan dipahami oleh manusia.
      </p>
      <p style="text-align: justify; margin-bottom: 10px;">
        Bahasa pemrograman tingkat tinggi (High-Level) maksudnya adalah bahasa yang semakin menjauh (sulit dimengerti) oleh komputer, tetapi semakin mendekat (mudah dimengerti) oleh manusia.
      </p>
      <p style="text-align: justify; margin-bottom: 10px;">
        Filosofi utamanya sangat sederhana: kode program harusnya mudah dimengerti, sejelas membaca kalimat dalam bahasa Inggris.
      </p>
      <p style="text-align: justify; margin-bottom: 10px;">
        Jika bahasa pemrograman lain memaksa penggunanya untuk menulis simbol-simbol rumit hanya untuk menampilkan teks, Python memangkas kerumitan tersebut. Ini membuat kita bisa lebih fokus pada pemecahan masalah (logika) daripada pusing memikirkan aturan penulisan kode (sintaks).
      </p>
    </td>
</table>

      <h3 style="margin-top: 0;">Bukti Nyata: Python vs Bahasa Lain</h2>
      <p style="text-align: justify; margin-bottom: 10px;">
        Teori saja tidak cukup. Mari kita lihat bukti langsung kenapa Python disebut "mudah dibaca".
      </p>
      <p style="text-align: justify; margin-bottom: 15px;">
        Misalnya, kita ingin menyuruh komputer melakukan tugas yang paling sederhana yaitu menampilkan teks <b>"Halo Dunia"</b> di layar.
      </p>
      <p style="margin-bottom: 5px;"><strong>1. Menggunakan Bahasa C</strong></p>
      <div style="background-color: #f0f0f0; padding: 10px; border-radius: 5px; border: 1px solid #ccc; font-family: 'Consolas', 'Monaco', monospace; font-size: 14px; margin-bottom: 15px; color: #333;">
        #include &lt;stdio.h&gt;<br><br>
        int main() {<br>
        &nbsp;&nbsp;printf("Halo Dunia");<br>
        &nbsp;&nbsp;return 0;<br>
        }
      </div>
      <p style="margin-bottom: 5px;"><strong>2. Menggunakan Python</strong></p>
      <div style="background-color: #e8f4fd; padding: 10px; border-radius: 5px; border: 1px solid #bbdefb; font-family: 'Consolas', 'Monaco', monospace; font-size: 14px; margin-bottom: 15px; color: #333;">
        print("Halo Dunia")
      </div>
      <p style="text-align: justify; margin-bottom: 10px;">
        Terlihat jelas perbedaannya. Di bahasa C, kamu harus menyiapkan "wadah" (include, main, return) dan memperhatikan detail kecil seperti titik koma (<code>;</code>) dan kurung kurawal (<code>{}</code>).
      </p>
      <p style="text-align: justify; margin-bottom: 0;">
        Sedangkan di Python, kamu berbicara langsung ke intinya. Tidak ada basa-basi, tidak ada simbol yang membingungkan. Inilah alasan kenapa Python sangat ramah untuk pemula.
      </p>
    </td>

    <td style="background-color: #ffffff; color: #000000; padding: 20px; border-radius: 5px;">
      <h3 style="margin-top: 0;">Cara Kerja: Interpreted vs Compiled</h3>
      <div align="center" style="margin-bottom: 20px;">
        <img src="https://i.pinimg.com/736x/2d/90/60/2d906036f1c6213ef723d0b61721b9a0.jpg" width="100%" style="max-width: 400px; border-radius: 10px;">
        <br>
        <small><i>Python mengeksekusi kode baris demi baris secara langsung.</i></small>
      </div>
      <p style="text-align: justify; margin-bottom: 15px;">
        Secara teknis, Python adalah bahasa <strong>Interpreted</strong>. Ini berbeda dengan bahasa C/C++ yang bersifat <strong>Compiled</strong>. Perbedaan utamanya ada pada cara kode diterjemahkan agar dimengerti komputer.
      </p>
      <div style="background-color: #fff3e0; padding: 15px; border-radius: 5px; border-left: 5px solid #ff9800; margin-bottom: 15px;">
        <strong>1. Tipe Compiled (C/C++, Java)</strong><br><br>
        Seluruh kode program harus diterjemahkan (compile) sekaligus menjadi file mesin (seperti (.exe)) sebelum bisa dijalankan.
        <br><br>
        <em>Karakteristik: Jika ada satu saja kesalahan sintaks, program tidak akan bisa dibuat (compile error) sampai semua diperbaiki.</em>
      </div>
      <div style="background-color: #e8f5e9; padding: 15px; border-radius: 5px; border-left: 5px solid #4caf50; margin-bottom: 15px;">
        <strong>2. Tipe Interpreted (Python, JavaScript)</strong><br><br>
        Kode program diterjemahkan dan dijalankan <strong>baris demi baris</strong> secara langsung tanpa membuat file mentahan (.exe) baru.
        <br><br>
        <em>Karakteristik: Program akan jalan dari atas. Jika ada error di baris ke-10, program akan berjalan normal dari baris 1-9, lalu berhenti tepat di baris 10.</em>
      </div>
      <p style="text-align: justify; margin-bottom: 0;">
        Sifat interpreted ini sangat membantu pemula. Kamu bisa langsung melihat hasil kode (output) atau pesan error secara instan tanpa proses kompilasi yang memakan waktu.
      </p>

      <h3 style="margin-top: 0; color: #000;">Tiga Karakteristik Python</h3>
      <p style="text-align: justify; margin-bottom: 20px; color: #333;">
        Sebelum menutup materi ini, ada tiga sifat teknis Python yang harus kamu ingat baik-baik. Tiga hal ini sering menjadi penyebab utama kebingungan bagi pemula.
      </p>
      <h4 style="margin-bottom: 5px; color: #1565c0;">1. Dynamic Typing (Otomatis)</h4>
      <p style="text-align: justify; margin-bottom: 10px; font-size: 14px; color: #333;">
        Di bahasa lama (Java/C++), kamu harus memberi tahu komputer tipe datanya dulu (misal: "Ini adalah angka"). Di Python, kamu tidak perlu melakukannya. Python cukup pintar untuk menebak isi variabelmu.
      </p>
      <div style="background-color: #f5f5f5; padding: 10px; border-radius: 5px; border: 1px solid #ccc; font-family: 'Consolas', 'Monaco', monospace; font-size: 14px; margin-bottom: 20px; color: #333;">
        <span style="color: #008000;"># Bahasa Lain (Ribet)</span><br>
        int umur = 57;<br>
        String nama = "Bahlil";<br><br>
        <span style="color: #0000ff;"># Python (Simpel)</span><br>
        umur = 57<br>
        nama = "Bahlil"
      </div>
      <h4 style="margin-bottom: 5px; color: #1565c0;">2. Indentasi</h4>
      <p style="text-align: justify; margin-bottom: 10px; font-size: 14px; color: #333;">
        Ini adalah aturan terpenting. Python tidak menggunakan kurung kurawal <code>{}</code> untuk mengelompokkan kode, melainkan menggunakan <strong>Spasi/Tab (Indentasi)</strong>.
      </p>
      <div style="background-color: #fff3e0; padding: 10px; border-radius: 5px; border-left: 5px solid #ff9800; margin-bottom: 20px; font-size: 14px; color: #333;">
        <strong>Penting:</strong> Salah menaruh spasi di awal baris bisa menyebabkan program <em>Error</em> dan tidak mau jalan. Kerapian di Python bukan cuma soal estetika, tapi soal aturan sintaks.
      </div>
      <h4 style="margin-bottom: 5px; color: #1565c0;">3. Case Sensitive</h4>
      <p style="text-align: justify; margin-bottom: 10px; font-size: 14px; color: #333;">
        Python membedakan huruf besar dan kecil. Variabel bernama <code style="background: #eee; padding: 5px; border-radius: 3px; border: 1px solid #ddd; color: #333;">nama</code> dianggap berbeda dengan <code style="background: #eee; padding: 5px; border-radius: 3px; border: 1px solid #ddd; color: #333;">Nama</code>
      </p>
      <div style="display: flex; gap: 10px; margin-bottom: 20px;">
        <code style="background: #eee; padding: 5px; border-radius: 3px; border: 1px solid #ddd; color: #333;">print()</code> &rarr; Benar ✅<br>
        <code style="background: #eee; padding: 5px; border-radius: 3px; border: 1px solid #ddd; color: #333;">Print()</code> &rarr; Salah ❌
      </div>
